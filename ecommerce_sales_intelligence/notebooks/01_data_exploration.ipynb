## Data Cleaning Decisions

Based on the raw-data investigation:

1. Remove duplicate transactions.
2. Exclude cancelled invoices where InvoiceNo starts with "C".
3. Exclude negative quantities from normal invoices because they represent
   damages, adjustments, or other non-sales stock movements.
4. Exclude transactions with UnitPrice <= 0 from sales analysis.
5. Keep transactions with missing CustomerID for overall sales analysis.
6. Handle missing CustomerID separately for customer-level analysis.
7. Keep missing Description values unless they prevent a specific analysis.
8. Keep the original raw dataset unchanged.

```python
# Create a working copy of the raw dataset
df_clean = df.copy()

print("Original shape:", df.shape)
print("Working copy shape:", df_clean.shape)
```

```python
# Remove duplicate rows

before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print("Rows before removing duplicates:", before)
print("Rows removed:", before - after)
print("Rows after removing duplicates:", after)
```

```python
# Remove cancelled invoices

before = len(df_clean)

cancelled = df_clean["InvoiceNo"].astype(str).str.startswith("C")

df_clean = df_clean[~cancelled]

after = len(df_clean)

print("Rows before removing cancellations:", before)
print("Cancelled rows removed:", before - after)
print("Rows after removing cancellations:", after)
```

```python
# Remove remaining negative-quantity transactions

before = len(df_clean)

negative_quantity = df_clean["Quantity"] < 0

df_clean = df_clean[~negative_quantity]

after = len(df_clean)

print("Rows before removing negative quantities:", before)
print("Negative-quantity rows removed:", before - after)
print("Rows after removing negative quantities:", after)
```

```python
# Check remaining invalid prices

zero_price = (df_clean["UnitPrice"] == 0)
negative_price = (df_clean["UnitPrice"] < 0)

print("Remaining zero-price rows:", zero_price.sum())
print("Remaining negative-price rows:", negative_price.sum())
print("Total invalid-price rows:", (zero_price | negative_price).sum())
```

```python
# Remove transactions with invalid prices

before = len(df_clean)

invalid_price = df_clean["UnitPrice"] <= 0

df_clean = df_clean[~invalid_price]

after = len(df_clean)

print("Rows before removing invalid prices:", before)
print("Invalid-price rows removed:", before - after)
print("Rows after removing invalid prices:", after)
```

```python
# Final validation of transaction data

print("Final shape:", df_clean.shape)

print("\nNegative quantities:",
      (df_clean["Quantity"] < 0).sum())

print("Zero quantities:",
      (df_clean["Quantity"] == 0).sum())

print("Zero prices:",
      (df_clean["UnitPrice"] == 0).sum())

print("Negative prices:",
      (df_clean["UnitPrice"] < 0).sum())

print("Duplicate rows:",
      df_clean.duplicated().sum())

print("Cancelled invoices:",
      df_clean["InvoiceNo"].astype(str).str.startswith("C").sum())
```

```python
# Check missing values after cleaning

missing_clean = df_clean.isnull().sum()
missing_percent_clean = (missing_clean / len(df_clean)) * 100

print("Missing values:")
print(missing_clean)

print("\nMissing percentage:")
print(missing_percent_clean.round(2))
```

```python
# Check missing CustomerID and Description together

print("Missing CustomerID:",
      df_clean["CustomerID"].isnull().sum())

print("Missing Description:",
      df_clean["Description"].isnull().sum())

print("Total rows:",
      len(df_clean))
```

```python
# Create dataset for customer-level analysis

df_customer = df_clean.dropna(subset=["CustomerID"]).copy()

print("Customer analysis dataset:", df_customer.shape)
print("Missing CustomerID:", df_customer["CustomerID"].isnull().sum())
```

```python
# Save cleaned datasets

df_clean.to_csv("../data/processed/clean_sales.csv", index=False)
df_customer.to_csv("../data/processed/customer_sales.csv", index=False)

print("Processed datasets saved successfully!")
```
