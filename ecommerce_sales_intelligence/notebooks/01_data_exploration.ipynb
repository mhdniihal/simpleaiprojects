```python
import pandas as pd
import numpy as np

df = pd.read_excel("../data/raw/Online Retail.xlsx")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
print("Columns:")
for column in df.columns:
    print("-", column)
    
```

```python
df.info()
```

```python
df.describe(include="all")
```

```python
missing = df.isnull().sum()

print("Missing values in each column:")
print(missing)
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

print("Missing percentage:")
print(missing_percent.round(2))
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
print("Quantity statistics:")
print(df["Quantity"].describe())

print("\nUnitPrice statistics:")
print(df["UnitPrice"].describe())
```

```python
print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Zero quantities:", (df["Quantity"] == 0).sum())

print("Zero prices:", (df["UnitPrice"] == 0).sum())
print("Negative prices:", (df["UnitPrice"] < 0).sum())
```

```python
# Check cancelled invoices
cancelled_invoices = df["InvoiceNo"].astype(str).str.startswith("C")

print("Cancelled invoice rows:", cancelled_invoices.sum())
print("Normal invoice rows:", (~cancelled_invoices).sum())
```

```python
# Compare quantities in cancelled vs normal invoices

print("Negative quantities in cancelled invoices:",
      ((df["Quantity"] < 0) & cancelled_invoices).sum())

print("Negative quantities in normal invoices:",
      ((df["Quantity"] < 0) & (~cancelled_invoices)).sum())
```

```python
# Look at some cancelled transactions

df[cancelled_invoices].head(10)
```

```python
# Negative quantities that are NOT marked as cancelled
unusual_negative = df[
    (df["Quantity"] < 0) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))
]

print("Unusual negative-quantity rows:", len(unusual_negative))

unusual_negative.head(20)
```

```python
# Check their invoice numbers and descriptions

print("Number of unique invoices:",
      unusual_negative["InvoiceNo"].nunique())

print("\nSample invoice numbers:")
print(unusual_negative["InvoiceNo"].unique()[:20])
```

```python
# Check whether these transactions have unusual descriptions

print(
    unusual_negative["Description"]
    .value_counts()
    .head(20)
)
```

```python
# Investigate zero-price transactions

zero_price = df[df["UnitPrice"] == 0]

print("Zero-price rows:", len(zero_price))
print("Unique invoices:", zero_price["InvoiceNo"].nunique())
print("Unique customers:", zero_price["CustomerID"].nunique())

zero_price.head(20)
```

```python
# Check descriptions of zero-price transactions

print(
    zero_price["Description"]
    .value_counts()
    .head(20)
)
```

```python
# Check quantities in zero-price transactions

print(zero_price["Quantity"].describe())

print("\nPositive quantities:", (zero_price["Quantity"] > 0).sum())
print("Negative quantities:", (zero_price["Quantity"] < 0).sum())
```
